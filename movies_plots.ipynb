```python
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv(r'C:\Users\lohit\OneDrive\Documents\top_20_indian_movies_collection-1.csv')
df
```

```python
plt.figure(figsize=(12,5))

plt.plot(df['movie_name'],df['release_year'],linestyle='--',marker='o')

plt.xlabel('movies names')
plt.ylabel('release year')
plt.title('movies release years')

print(plt.style.available)
plt.style.use('dark_background')

plt.tight_layout()
plt.xticks(rotation=90)
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(12,5))

plt.bar(df['movie_name'],df['worldwide_collection_inr_crore'])
plt.plot(df['movie_name'],df['worldwide_collection_inr_crore'],color='orange',linestyle='--',marker='o')

plt.xlabel('movies names')
plt.ylabel('collections')
plt.title('highest collection of Indian movies')

print(plt.style.available)
plt.style.use('dark_background')

plt.tight_layout()
plt.xticks(rotation=90)
plt.grid(True)
plt.show()
```

```python
language_count=df['language'].value_counts()
explode=[0,0,0,0.4]

plt.pie(language_count,labels=language_count.index,explode=explode,autopct='%1.1f%%',shadow=True,wedgeprops={'edgecolor':'black'})
plt.style.use('seaborn-v0_8-talk')

plt.xticks(rotation=90)
plt.title('numbers languages in highset collected movies ')
```

```python
plt.figure(figsize=(12,5))

plt.scatter(df['movie_name'],df['release_year'],color='red',linestyle='--',marker='o')

plt.xlabel('movies names')
plt.ylabel('release year')
plt.title('movies release years')

print(plt.style.available)
plt.style.use('dark_background')

plt.tight_layout()
plt.xticks(rotation=90)
plt.grid(True)
plt.show()
```
