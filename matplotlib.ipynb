```python
import matplotlib
import matplotlib.pyplot as plt

x=[20,23,24,22,21,34,24]

y1=[3400,5000,6000,7000,3300,9000,6000]
y2=[9000,4000,5000,3300,2220,30000,40000]

plt.plot(x,y1,color='b',marker='o',label='salary of Java')
plt.plot(x,y2,color='k',linestyle='--',marker='*',label='salary of python')

plt.xlabel('age of employess')
plt.ylabel('salaries')
plt.title('salary of employess by there ages')

plt.tight_layout()
plt.legend()
plt.grid()

print(plt.style.available)
plt.style.use('seaborn-v0_8-whitegrid')

plt.show()
```

```python
import matplotlib
import matplotlib.pyplot as plt
import numpy as np

x=[20,23,24,22,21,34,24]
x_indexes=np.arange(len(x))
width=0.5

y1=[3400,5000,6000,7000,3300,9000,6000]
y2=[9000,4000,5000,3300,2220,30000,40000]
y3=[400,555,66666,444,333,222,555]

plt.bar(x_indexes -width,y1,width=width,color='b',label='salary of Java')
plt.bar(x_indexes,y2,width=width,color='k',label='salary of python')
plt.bar(x_indexes +width,y3,width=width,color='r',label='salary of SQL')

plt.xlabel('age of employess')
plt.ylabel('salaries')
plt.title('salary of employess by there ages')

plt.tight_layout()
plt.legend()
plt.grid(True)
plt.xticks(ticks=x_indexes,label=x)

print(plt.style.available)
plt.style.use('classic')

plt.show()
```

```python
import matplotlib
import matplotlib.pyplot as plt

x=[20,30,22,23,25]

y1=[200,222,200,300,400]
y2=[110,200,400,500,600]
y3=[600,500,400,300,200]

plt.plot(x,y1,color='r',marker='o',label='salary of teacher')
plt.plot(x,y2,color='k',linestyle='--',marker='*',label='salary of engineering')
plt.plot(x,y3,color='b',label='salary of Doctor')

plt.grid(True)
plt.tight_layout()
plt.legend()
plt.show()

```

```python
import matplotlib
import matplotlib.pyplot as plt

x=[20,30,22,23,25]
index_x=np.arange(len(x))
width=0.25

y1=[200,222,200,300,400]
y2=[110,200,400,500,600]
y3=[600,500,400,300,200]

plt.bar(index_x -width,y1,width=width,color='r',label='salary of teacher')
plt.bar(index_x,y2,width=width,color='k',label='salary of engineering')
plt.bar(index_x +width,y3,width=width,color='b',label='salary of Doctor')

plt.grid(True)
plt.tight_layout()
plt.legend(fontsize=10,loc='upper left')
plt.xticks(ticks=index_x,labels=x)
plt.show()
```

```python
import matplotlib
import matplotlib.pyplot as plt

x=[20,30,22,23,25]
labels=['jave','C++','python','SQL','DSA']
explode=[0,0,0,0.2,0]

plt.pie(x,labels=labels,explode=explode,startangle=90,autopct='%1.1f%%',shadow=True,wedgeprops={'edgecolor':'black'})

plt.title('my pie chart')
plt.tight_layout()
plt.legend(fontsize=8)
plt.show
```

```python
import matplotlib
import matplotlib.pyplot as plt

x=[20,30,22,23,25]
labels=['java','C','C++','SQL','php']
explode=[0,0,0,0,0.5]
colors=('red','orange','yellow','green','violet')

plt.pie(x,labels=labels,explode=explode,colors=colors,shadow=True,autopct='%1.1f%%',wedgeprops={'edgecolor':'black'})

plt.title('my pie chart')
plt.tight_layout()
plt.legend(fontsize=8,loc='upper left')
plt.show
```
